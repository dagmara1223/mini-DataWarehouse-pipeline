**Data Type Validation**

```python
import pandas as pd

df = pd.read_csv('/content/sample_data/part2.csv', index_col=['Unnamed: 0'])
df_copy = df.copy()
```

```python
df_copy.head()
```

Firstly, let's display each column type

```python
df_copy.dtypes
```

**1. Date types** <br>
My first suspect is 'date' column that for now stores object type. Type should be converted into DataType.

```python
df_copy['date'] = pd.to_datetime(df_copy['date']).dt.date
type(df_copy['date'].iloc[0])
```

**2. Numeric Types** <br>
Let's review numeric values.

```python
df_copy.dtypes
```

```python
df_copy.head()
```

All numeric values are stored correctly.

**3. Memory & efficiency improvement - Object -> Categorical values** <br>
Firstly let's show how many unique values each column contains.

```python
df_copy.nunique()
```

Columns with a relatively small number of unique values are good candidates for conversion to the categorical data type.

Converting such columns to category can significantly reduce memory usage and improve performance, especially when working with large datasets. This optimization is particularly useful when columns represent repeated labels or discrete categories rather than free-text values. My main suspects are : status, fulfilment, ship-service-level, category, size, courier status, currency, ship-state, ship-country

```python
df_copy['status'] = df_copy['status'].astype('category')
df_copy['fulfilment'] = df_copy['fulfilment'].astype('category')
df_copy['ship-service-level'] = df_copy['ship-service-level'].astype('category')
df_copy['category'] = df_copy['category'].astype('category')
df_copy['size'] = df_copy['size'].astype('category')
df_copy['courier status'] = df_copy['courier status'].astype('category')
df_copy['currency'] = df_copy['currency'].astype('category')
df_copy['ship-state'] = df_copy['ship-state'].astype('category')
df_copy['ship-country'] = df_copy['ship-country'].astype('category')
```

We may also see that column : 'b2b' contains only boolean values.

```python
df_copy['b2b'] = df_copy['b2b'].astype('boolean')
```
